```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
titanic = sns.load_dataset("titanic")
```

```python
titanic.head()
```

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(
    x="age",
    y="fare",
    hue="survived",
    data=titanic,
)
plt.title("Age vs Fare by Survival and Sex")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.show()
```

```python
plt.figure(figsize=(8, 6))
plt.bar(titanic.age, titanic.fare)
plt.title("Age vs Fare Trends by Passenger Class")
plt.show()
```

```python
plt.figure(figsize=(8, 6))
plt.scatter(
    titanic["age"], titanic["fare"], c=titanic["survived"], cmap="Blues", alpha=0.6
)
plt.colorbar(label="Servived")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.title("Matplotlib Scatter Plot: Age vs Fare")
plt.show()
```

```python
# Load dataset
titanic = sns.load_dataset("titanic")

print(titanic.columns)
corr_matrix = titanic[["survived", "age", "fare", "pclass"]].corr()

plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, fmt=".2f", linewidths=0.5)
plt.title("Titanic Dataset Correlation Heatmap")
plt.show()
```

```python
flights = sns.load_dataset("flights")
```

```python
sns.relplot(
    x="year",
    y="passengers",
    hue="month",
    kind="scatter",
    data=flights,
)
plt.title("Relational Line Plot for Flights")
plt.show()
```
