```python
import pandas as pd
from sklearn.datasets import load_iris
```

```python
data = load_iris()
df = pd.DataFrame(data.data)
columns = ["sepal_length", "sepal_width", "petal_length", "petal_width"]
df.columns  = columns
```

```python
cov_mat = [
    df.sepal_length.cov(df.sepal_width),
    df.sepal_length.cov(df.petal_length),
    df.sepal_length.cov(df.petal_width),
]

cor_mat = [
    df.sepal_length.corr(df.sepal_width),
    df.sepal_length.corr(df.petal_length),
    df.sepal_length.corr(df.petal_width),
]

cov_mat, cor_mat
```

```python
mean_mat = [
    df.sepal_length.mean(),
    df.sepal_width.mean(),
    df.petal_length.mean(),
    df.petal_width.mean(),
]

std_mat = [
    df.sepal_length.std(),
    df.sepal_width.std(),
    df.petal_length.std(),
    df.petal_width.std(),
]

mean_mat, std_mat
```
