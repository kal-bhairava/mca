### All possible **2D** pairs from `10 X 10` matrix

```python
np.random.seed(43)
array = np.random.rand(10, 10)

empty_list = []

for i in range(1, 101):
  if 100 % i == 0:
    empty_list.append(array.reshape(i, 100 // i))

len(empty_list)
```

### All possible **3D** pairs from `10 X 10` matrix

```python
empty_list = []

for i in range(1, 101):
  for j in range(1, 101):
    for k in range(1, 101):
      if (i * j * k) == 100:
        empty_list.append(array.reshape((i, j, k)))

len(empty_list)
```
