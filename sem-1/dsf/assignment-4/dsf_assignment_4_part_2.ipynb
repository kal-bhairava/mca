-  Create a NumPy array containing marks of 10 students in 5 subjects. Perform:

```python
marks = np.random.randint(50, 101, size=(10, 5))
print(marks)
```

1. Calculate subject-wise average.  

```python
subject_averages = np.mean(marks, axis=0)
print(f"Subject-wise Averages: {subject_averages}")
```

2. Calculate student-wise average.  

```python
student_averages = np.mean(marks, axis=1)
print(f"Student-wise Averages:\n{student_averages}")
```

3. Find highest and lowest marks.  

```python
highest_mark = np.max(marks)
lowest_mark = np.min(marks)
print(f"Highest Mark Overall: {highest_mark} | Lowest Mark Overall: {lowest_mark}")
```

4. Find students scoring above the overall average.  

```python
overall_average = np.mean(marks)
above_avg_indices = np.where(student_averages > overall_average)[0]
print(f"Overall Average Mark: {overall_average:.2f}")
print(f"Students (0-indexed) scoring above overall average: {above_avg_indices}")
```

5. Find standard deviation for every subject.  

```python
subject_std = np.std(marks, axis=0)
print(f"Subject-wise Standard Deviation: {subject_std}")
```

6. Sort students according to their average marks.

```python
sorted_student_indices = np.argsort(student_averages)
print(f"Student indices sorted by performance (lowest to highest average): {sorted_student_indices}")
```

7. Identify the top 3 students.

```python
top_3_students = sorted_student_indices[-3:][::-1]
print(f"Top 3 Student Indices (highest to lowest): {top_3_students}")
print(f"Top 3 Student Averages: {student_averages[top_3_students]}")
```

# Q. 7
- Create a NumPy array containing salaries of employees.

```python
np.random.seed(101)
salaries = np.random.randint(25000, 120000, size=15)
```

Mean salary  

```python
mean_salary = np.mean(salaries)
print(f"Mean Salary: ₹{mean_salary:,.2f}")
```

Median salary  

```python
median_salary = np.median(salaries)
print(f"Median Salary: ₹{median_salary:,.2f}")
```

Minimum salary  

```python
max_salary = np.max(salaries)
print(f"Maximum Salary: ₹{max_salary:,.2f}")
```

Minimum salary

```python
min_salary = np.min(salaries)
print(f"Minimum Salary: ₹{min_salary:,.2f}")
```

Standard deviation  

```python
std_salary = np.std(salaries)
print(f"Standard Deviation: ₹{std_salary:,.2f}")
```

Employees earning above average  

```python
above_avg_salaries = salaries[salaries > mean_salary]
print(f"Salaries above average (₹{mean_salary:,.2f}): {above_avg_salaries}")
```

Employees earning between ₹40,000 and ₹80,000  

```python
between_40k_80k = salaries[(salaries >= 40000) & (salaries <= 80000)]
print(f"Employees earning between ₹40,000 and ₹80,000: {between_40k_80k}")
```

Percentage of employees earning above ₹60,000  

```python
above_60k_count = np.sum(salaries > 60000)
total_employees = salaries.size
percentage_above_60k = (above_60k_count / total_employees) * 100
print(f"Percentage earning above ₹60,000: {percentage_above_60k:.2f}% ({above_60k_count}/{total_employees} employees)")
```

# Q 9
- Create two NumPy matrices: A = 3 × 3 B = 3 × 3

```python
np.random.seed(24)
A = np.random.randint(1, 10, size=(3, 3))
B = np.random.randint(1, 10, size=(3, 3))
```

1. Addition  

```python
add_result = A + B
print(f"Matrix Addition (A + B):\n{add_result}\n")
```

2. Subtraction  

```python
sub_result = A - B
print(f"Matrix Subtraction (A - B):\n{sub_result}\n")
```

3. Element-wise multiplication  

```python
elem_mult = A * B
print(f"Element-wise Multiplication (A * B):\n{elem_mult}\n")
```

4. Matrix multiplication  

```python
matrix_mult = A @ B
print(f"Matrix Multiplication (A @ B):\n{matrix_mult}\n")
```

5. Transpose  

```python
transpose_A = A.T
print(f"Transpose of Matrix A (A.T):\n{transpose_A}\n")
```

6. Row-wise sum  

```python
row_sum_A = np.sum(A, axis=1)
print(f"Row-wise Sum of Matrix A: {row_sum_A}\n")
```

7. Column-wise sum

```python
col_sum_A = np.sum(A, axis=0)
print(f"Column-wise Sum of Matrix A: {col_sum_A}\n")
```

8. Maximum element  

```python
max_A = np.max(A)
print(f"Maximum Element in Matrix A: {max_A}\n")
```

9. Minimum element  

```python
min_A = np.min(A)
print(f"Minimum Element in Matrix A: {min_A}\n")
```

10. Calculate determinant  

```python
det_A = np.linalg.det(A)
print(f"Determinant of Matrix A: {det_A:.4f}\n")
```

11. Calculate inverse, if possible  

```python
if det_A != 0:
    inverse_A = np.linalg.inv(A)
    print(f"Inverse of Matrix A (A^-1):\n{inverse_A}\n")

    print("Verification (A @ A^-1 ≈ Identity Matrix):")
    print(np.round(A @ inverse_A))
else:
    print("Inverse of Matrix A is not possible (Matrix is Singular / Det = 0)\n")
```

# Q 11
- Create a NumPy array containing electricity consumption for 12 months.

```python
np.random.seed(42)
monthly_consumption = np.random.randint(150, 450, size=12)
months = np.array(["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"])

for m, c in zip(months, monthly_consumption):
    print(f"{m}: {c} kWh")
```

1. Annual consumption

```python
annual_total = np.sum(monthly_consumption)
print(f"Total Annual Consumption: {annual_total} kWh")
```

2. Monthly average  

```python
monthly_avg = np.mean(monthly_consumption)
print(f"Monthly Average Consumption: {monthly_avg:.2f} kWh")
```

3. Highest consumption month  

```python
max_index = np.argmax(monthly_consumption)
print(f"Highest Consumption Month: {months[max_index]} ({monthly_consumption[max_index]} kWh)")
```

4. Lowest consumption month  

```python
min_index = np.argmin(monthly_consumption)
print(f"Lowest Consumption Month: {months[min_index]} ({monthly_consumption[min_index]} kWh)")
```

5. Months above average  

```python
above_avg_mask = monthly_consumption > monthly_avg
print(f"Months scoring Above Average ({monthly_avg:.2f} kWh):")
print(f"Months: {months[above_avg_mask]}")
print(f"Values: {monthly_consumption[above_avg_mask]}")
```

6. Standard deviation  

```python
std_dev = np.std(monthly_consumption)
print(f"Standard Deviation: {std_dev:.2f} kWh")
```

7. Percentage contribution of each month  

```python
percentage_contribution = (monthly_consumption / annual_total) * 100
print("Percentage Contribution of Each Month:")
for m, p in zip(months, percentage_contribution):
    print(f"{m}: {p:.2f}%")
```

8. Sort months according to consumption

```python
sorted_indices = np.argsort(monthly_consumption)
print("Months Sorted by Consumption (Lowest to Highest):")
print(f"Months: {months[sorted_indices]}")
print(f"Values: {monthly_consumption[sorted_indices]}")
```

# Q 10
- Create a NumPy matrix representing grayscale pixel values from 0–255.

```python
np.random.seed(42)
image_matrix = np.random.randint(0, 256, size=(6, 6))

print(image_matrix)
```

1. Find minimum and maximum pixel values.  

```python
min_pixel = np.min(image_matrix)
max_pixel = np.max(image_matrix)
print(f"Minimum Pixel Value: {min_pixel} | Maximum Pixel Value: {max_pixel}\n")
```

2. Calculate mean intensity.  

```python
mean_intensity = np.mean(image_matrix)
print(f"Mean Intensity: {mean_intensity:.2f}\n")
```

3. Convert the image to binary using a threshold.  

```python
threshold = 127
binary_image = (image_matrix > threshold).astype(int)
print(f"Binary Image (Threshold = {threshold}):\n{binary_image}\n")
```

4. Extract a specific region using slicing.  

```python
central_patch = image_matrix[1:4, 1:4]
print(f"Extracted Central Patch (Rows 1-3, Cols 1-3):\n{central_patch}\n")
```

5. Flip the matrix horizontally and vertically.  

```python
flipped_horizontal = image_matrix[:, ::-1]
flipped_vertical = image_matrix[::-1, :]

print(f"Horizontally Flipped Matrix:\n{flipped_horizontal}\n")
print(f"Vertically Flipped Matrix:\n{flipped_vertical}\n")
```

6. Calculate row-wise and column-wise intensity.  

```python
row_means = np.mean(image_matrix, axis=1)
col_means = np.mean(image_matrix, axis=0)

print(f"Row-wise Mean Intensities: {np.round(row_means, 2)}")
print(f"Column-wise Mean Intensities: {np.round(col_means, 2)}")
```
