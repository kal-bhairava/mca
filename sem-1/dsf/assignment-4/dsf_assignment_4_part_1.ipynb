2. Display all the records

```python
data = []
for l in reader:
  data.append(l)
```

3. Calculate average marks of each student

4. Find student with highest average

```python
ham = 1 # assuming first student has the highest average/mean
for i,rec in enumerate(data[1:]):
  ham = i if data[ham][-1] < rec[-1] else ham
```

```python
ham
```

```python
data[ham]
```

5. Hight and lowest marks in each subject

```python
min = [100, 100, 100, 100]
for rec in data[1:]:
  for idx,marks in enumerate(rec[2:-3]):
    if marks == '': # some records do not have any value
      continue
    marks = int(marks)
    min[idx] = marks if min[idx] > marks else min[idx]

min
```

6. Identify students with `> 75 %` attendance

```python
studs = []
for idx,rec in enumerate(data[1:]):
  if int(rec[-2]) > 75:
    studs.append(idx+1) # + 1 as the first one is column name / label
```

```python
studs[:10]
```

7. Identify students who `passed` or `failed` given mean is `>=40`

```python
passed = []
for idx,rec in enumerate(data[1:]):
  if int(rec[-1]) >= 40:
    passed.append(idx+1)
```

```python
passed[:10]
```

```python
for i in passed[:10]:
  print(data[i][1])
```

8. Write final analysis to new csv file

```python
keys = [
    "Roll No.",
    "Name",
    "Sub 1",
    "Sub 2",
    "Sub 3",
    "Sub 4",
    'Attendance',
    'Mean',
    'Passed',
    'Attendanced >= 75%',
]

final_data = []

for idx, student in enumerate(data[1:], start=1):

  student_row = list(student)

  passed_flag = 1 if idx in passed else 0
  attendance_flag = 1 if idx in studs else 0

  student_row.append(passed_flag)
  student_row.append(attendance_flag)

  final_data.append(student_row)


output_filename = "final_student_analysis.csv"
with open(output_filename, "w") as f:
  writer = csv.writer(f)

  writer.writerow(keys),
  writer.writerows(final_data)
```

### 2. Employee Salaey Analysis

1. read csv file

```python
!wget https://raw.githubusercontent.com/kal-bhairava/mca/main/dataset/assignment/sem-1/dsf/employee_data.csv

dataset = []

with open('employee_data.csv', 'r') as file:
  reader = csv.reader(file)
  dataset.extend(list(reader))

dataset[:3]
```

```python
import copy

# parsing
keys = copy.deepcopy(dataset[0])
dataset = list(map(lambda rec : [rec[0], rec[1], float(rec[2]), float(rec[3]), rec[4], int(rec[5])], dataset[1:]))
```

```python
print(keys)
```

2. Calculate total salary

```python
keys = keys.append('total')
for idx,rec in enumerate(dataset):
  total = rec[2] + rec[3]
  dataset[idx].append(total)

dataset[:2]
```

3. Highest paid employee

```python
emp = 1
for idx,rec in enumerate(dataset):
  if(rec[-1] > dataset[emp][-1]):
    emp = idx

dataset[emp]
```

4. Department wise avg salary

```python
dept = {}
for idx,rec in enumerate(dataset):
  departement = rec[4]
  salary = rec[-1]
  count = dept[departement][1] + 1 if departement in dept else 1
  if departement in dept:
    salary = ((dept[departement][0] * count )+ rec[-1]) / count
    dept[departement] = [salary, count]
  else:
    dept[departement] = [salary, count]
```

```python
dept
```

5. Employee with 5+ years of experience

```python
emp = []
for idx,rec in enumerate(dataset):
  if(dataset[idx][5] > 5):
    emp.append(idx)

emp[:10]
```

6. Sort employees accoarding to salary

```python
sorted_employees = []

for employee in dataset:
  inserted = False
  for idx,sorted in enumerate(sorted_employees):
    if(sorted[-1] < employee[-1]):
      sorted_employees.insert(idx, employee)
      break
  if not inserted:
    sorted_employees.append(employee)

sorted_employees[:5]
```

7. Store the sorted daat into csv

```python
print(keys)
```

```python
with open('sorted_employee_data.csv', "w", newline="", encoding="utf-8") as file:
  writer = csv.writer(file)
  writer.writerows(sorted_employees)
```

### 3 Sales data analysis
1. Calculate total sales for every order.

```python
!wget https://raw.githubusercontent.com/kal-bhairava/mca/main/dataset/assignment/sem-1/dsf/sales_data.csv
```

```python
sales_data = open('sales_data.csv', 'r')
sales_data_reder = csv.reader(sales_data)
sales_dataset = list(sales_data_reder)
sales_dataset[:2]
```

```python
#parsing
sales = list(map(lambda rec : [rec[0], int(rec[1]), rec[2], rec[3], float(rec[4]), float(rec[5])], sales_dataset[1:]))
```

```python
for rec in sales:
  rec.append(rec[-1] * rec[1])

sales[:3]
```

3. Calculating region wise revenue

```python
revenue = {}

for rec in sales:
  region = rec[3]
  if region in revenue:
    revenue[region] += rec[-1]
  else:
    revenue[region] = rec[-1]

revenue
```

4. top 5 orders by revenue

```python
sorted_sales = sorted(sales, key=lambda rec : rec[-1])
sorted_sales[:5]
```

```python
with open('summarized_sales.csv', mode='w', newline="", encoding="utf-8") as file:
  writer = csv.writer(file)
  writer.writerows(revenue)
```

### 4 Weather Data Processing

```python
!wget https://raw.githubusercontent.com/kal-bhairava/mca/main/dataset/assignment/sem-1/dsf/movie_data.csv
```

```python
movie_data = open('movie_data.csv', 'r')
movie_data_reder = csv.reader(movie_data)
movie_dataset = list(movie_data_reder)
movie_dataset[:2]
```

```python
#parsing
movies = list(map(lambda rec : [int(rec[0]), int(rec[1]), rec[2], rec[3], int(rec[4])], movie_dataset[1:]))
```

1. Highest rated movie

```python
sorted_movies = sorted(movies, key=lambda rec : [rec[-1], [1]], reverse=[True, True])
```

```python
sorted_movies[:1]
```

2. Top 10 movies based on rating

```python
sorted_movies[:11]
```

3. Calculate average rating by genere

```python
mean_rating = {}
for movie in movies:
  genere = movie[-2]
  rating = movie[-1]

  rating = float(rating) if rating != '' else 0.0
  if genere in mean_rating:
    current_sum = mean_rating[genere][0]
    current_count = mean_rating[genere][1]

    mean_rating[genere] = [current_sum + rating, current_count + 1]
  else:
    mean_rating[genere] = [rating, 1]

final_averages = {}
for genere, (total_rating, count) in mean_rating.items():
  final_averages[genere] = round(total_rating / count, 2)

final_averages
```

4. Find the number of movies released each year.  

```python
movies_relased_each_year = {}
for movie in movies:
  year = movie[1]

  if year in movies_relased_each_year:
    movies_relased_each_year[year] += 1
  else:
    movies_relased_each_year[year] = 1
```

5. Find genres having more than 20 movies.  

```python
gen = []
for genere in mean_rating.keys():
  if(mean_rating[genere][1] > 20):
    gen.append(genere)

gen
```

6. Save the top-rated movies into a separate CSV file.

```python
with open('top_rated_movies', mode="w", newline="", encoding='utf-8') as file:
  writer = csv.writer(file)
  writer.writerow(movie_dataset[:1])
  writer.writerows([rec for rec in sorted_movies if float(rec[-1]) >= 4])
```
