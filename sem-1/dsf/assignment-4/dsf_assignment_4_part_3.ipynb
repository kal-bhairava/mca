# Q.11
### Importing **`Matpltlib`**

```python
import matplotlib.pyplot as plt
import numpy as np
```

1. Bar chart — average marks of students.  
2. Line graph — marks in different subjects.
3. Histogram — distribution of total marks.
4. Add title, labels and legend.
5. Display appropriate grid where required.  

```python
np.random.seed(42)
marks = np.random.randint(50, 101, size=(10, 5))

names = ["Liam Smith", "Sophia Jones", "Noah Williams", "Emma Brown", "Oliver Davis", "Ava Miller", "Elijah Wilson", "Isabella Moore", "Lucas Taylor", "Mia Anderson"]
subjects = ["C", "Cpp", "C#", "Java", "Python"]

student_averages = np.mean(marks, axis=1)

plt.figure(figsize=(13, 3))
plt.bar(names, student_averages)
plt.title("Average Marks")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.grid(axis='y')
plt.show()
```

```python
plt.plot(subjects, marks.T)
plt.title("Subject Marks")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.grid(True)
plt.legend(names)
```

```python
plt.hist(marks.sum(axis=1))
plt.title("Total Marks Distribution")
plt.xlabel("Total Marks")
plt.ylabel("Number of Students")
plt.grid(axis='y')
```

# Q 13

- Line graph of temperature.
- Bar chart of rainfall.
- Histogram of temperature.
- Scatter plot of temperature vs. rainfall.
- Add appropriate titles, labels and legends.  

```python
np.random.seed(42)
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
temperature = np.random.randint(15, 38, size=12)
rainfall = np.random.randint(10, 200, size=12)
```

1 Line plot

```python
plt.figure()
plt.plot(months, temperature, marker='o', color='tab:red', label='Temperature (°C)')
plt.title('Monthly Temperature Trend')
plt.xlabel('Months')
plt.ylabel('Temperature (°C)')
plt.grid(True)
plt.legend()
plt.show()
```

2. Bar plot

```python
plt.figure()
plt.bar(months, rainfall, color='tab:blue', label='Rainfall (mm)')
plt.title('Monthly Rainfall Breakdown')
plt.xlabel('Months')
plt.ylabel('Rainfall (mm)')
plt.grid(axis='y')
plt.legend()
plt.show()
```

3. Histogram

```python
plt.figure()
plt.hist(temperature, bins=4, color='tab:orange', edgecolor='black', label='Temp Distribution')
plt.title('Distribution of Temperatures')
plt.xlabel('Temperature Range (°C)')
plt.ylabel('Frequency (Months)')
plt.grid(axis='y')
plt.legend()
plt.show()
```

4. Scatter plot

```python
plt.figure()
plt.scatter(temperature, rainfall, color='tab:green', s=100, edgecolor='black', label='Data Points')
plt.title('Temperature vs. Rainfall Correlation')
plt.xlabel('Temperature (°C)')
plt.ylabel('Rainfall (mm)')
plt.grid(True)
plt.legend()
plt.show()
```

1. Bar chart of employee salaries.
2. Histogram of salary distribution.  
3. Pie chart showing department-wise employee distribution.  
4. Box plot of salaries.
5. Add suitable titles and labels.

```python
!wget https://raw.githubusercontent.com/kal-bhairava/mca/main/dataset/assignment/sem-1/dsf/employee_data.csv
```

```python
import pandas as pd
df = pd.read_csv('/content/employee_data.csv')
```

1. Bar Chart of Employee Salaries

```python
plt.figure()
plt.bar(df['name'][:10], df['basic_salary'][:10])
plt.title("Employee Salaries")
plt.xlabel("Employee Name")
plt.ylabel("Basic Salary")
plt.show()
```

2. Histogram of Salary Distribution

```python
plt.figure()
plt.hist(df['basic_salary'])
plt.title("Salary Distribution")
plt.xlabel("Salary Range")
plt.ylabel("Number of Employees")
plt.show()
```

3. Pie Chart showing Department-wise Employee Distribution

```python
plt.figure()
dept_counts = df['department'].value_counts()
plt.pie(dept_counts, labels=dept_counts.index)
plt.title("Department Distribution")
plt.show()
```

4. Box Plot of Salaries

```python
plt.figure()
plt.boxplot(df['basic_salary'])
plt.title("Salary Box Plot")
plt.ylabel("Salary")
plt.show()
```
