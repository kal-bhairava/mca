```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("/content/drive/MyDrive/student_performance.csv")

print("Dataset loaded successfully!")
```

Q1. Import and Display Dataset

```python
print(df)
print(df.head(5))
print(df.tail(3))
```

Q2. Basic Information of DataFrame

```python
print(df.info())
print(df.shape)
print(df.size)
print(df.columns)
print(df.dtypes)
```

Q3. Create and Modify DataFrame

```python
df["Total_Marks"] = df["Python_Marks"] + df["AI_Marks"] + df["Network_Marks"]

df["Average_Marks"] = df["Total_Marks"] / 3

df["Result"] = df["Average_Marks"].apply(lambda x: "Pass" if x >= 40 else "Fail")

print(df)
```

Q4. Adding and Deleting Columns

```python
df["Performance_Level"] = df["Average_Marks"].apply(
    lambda x: "Excellent" if x >= 85 else "Good" if x >= 70 else "Needs Improvement"
)

df.drop("Age", axis=1, inplace=True)

print(df)
```

Q5. Statistical Analysis

```python
subjects = ["Python_Marks", "AI_Marks", "Network_Marks"]

print(df[subjects].sum())
print(df[subjects].mean())
print(df[subjects].min())
print(df[subjects].max())
print(df[subjects].median())
```

Q6. Aggregate Function – agg()

```python
result = df[subjects].agg(["min", "max", "mean", "std"])

print(result)
```

Q7. Percentage Change

```python
df = df.sort_values("Student_ID")

df["Percentage_Change"] = df["Python_Marks"].pct_change() * 100

print(df[["Student_ID", "Python_Marks", "Percentage_Change"]])
```

Q8. Sorting Using sort_values()

```python
print(df.sort_values("Python_Marks"))

print(df.sort_values("Average_Marks", ascending=False))

print(df.sort_values(["Department", "Python_Marks"], ascending=[True, False]))

print(df.nlargest(3, "Average_Marks"))
```

Q9. Sorting Using sort_index()

```python
df1 = df.set_index("Student_ID")

print(df1.sort_index(ascending=False))

df1 = df1.reset_index()

print(df1)
```

Q10. Data Extraction Using Relational Operators

```python
print(df[df["Python_Marks"] > 80])

print(df[df["Attendance"] >= 90])

print(df[df["AI_Marks"] < 70])

print(df[df["Network_Marks"] != 75])
```

Q11. Data Extraction Using Logical Operators

```python
print(df[(df["Python_Marks"] > 75) & (df["AI_Marks"] > 80)])

print(df[(df["Attendance"] > 90) | (df["Study_Hours"] > 5)])

print(df[(df["Department"] == "CSE") & (df["Average_Marks"] > 75)])

print(df[~(df["Department"] == "IT")])
```

Q12. Data Extraction Using iloc

```python
print(df.iloc[:5])

print(df.iloc[3:8, 2:6])

print(df.iloc[-4:, :3])

print(df.iloc[1, df.columns.get_loc("Python_Marks")])
```

Q16. Box Plot

Q18. Student Performance Analysis System

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("/content/drive/MyDrive/student_performance.csv")

print(df.info())
print(df.shape)

df["Total_Marks"] = df["Python_Marks"] + df["AI_Marks"] + df["Network_Marks"]

df["Average_Marks"] = df["Total_Marks"] / 3

print(df)

topper = df.loc[df["Average_Marks"].idxmax()]

print("Highest Average Marks:")
print(topper[["Student_ID", "Name", "Average_Marks"]])

department_avg = df.groupby("Department")["Average_Marks"].mean()

print("Department-wise Average Marks:")
print(department_avg)

selected = df[(df["Attendance"] >= 85) & (df["Average_Marks"] >= 75)]

print("Selected Students:")
print(selected)

sorted_df = df.sort_values("Average_Marks", ascending=False)

print("Sorted Students:")
print(sorted_df)

department_avg.plot(kind="bar", color=["blue", "green", "orange"])

plt.title("Department-wise Average Marks")
plt.xlabel("Department")
plt.ylabel("Average Marks")
plt.show()

plt.hist(df["Average_Marks"], bins=5, color="lightgreen", edgecolor="black")

plt.title("Distribution of Average Marks")
plt.xlabel("Average Marks")
plt.ylabel("Number of Students")
plt.show()

print(df[["Student_ID", "Name", "Department", "Total_Marks", "Average_Marks"]])
```
