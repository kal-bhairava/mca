### Question 1: Correlation-Based Feature Redundancy

```python
import pandas as pd

data = {
    'F1': [10, 20, 30, 40, 50],
    'F2': [12, 24, 36, 48, 60],
    'F3': [50, 40, 30, 20, 10],
    'F4': [15, 25, 20, 35, 30]
}

df = pd.DataFrame(data)

corr = df.corr()
print(corr)

threshold = 0.90

for i in range(len(corr.columns)):
    for j in range(i + 1, len(corr.columns)):
        if corr.iloc[i, j] > threshold:
            print(corr.columns[i], corr.columns[j])
```

Observation: F1 and F2 are highly positively correlated and may be redundant.

### Question 2: Distance-Based Feature Similarity

```python
import numpy as np

F1 = np.array([2, 4, 6, 8, 10])
F2 = np.array([3, 5, 7, 9, 11])

euclidean = np.linalg.norm(F1 - F2)
manhattan = np.sum(np.abs(F1 - F2))
cosine = np.dot(F1, F2) / (np.linalg.norm(F1) * np.linalg.norm(F2))

print("Euclidean Distance:", euclidean)
print("Manhattan Distance:", manhattan)
print("Cosine Similarity:", cosine)
```

### Question 4: Measuring Redundancy Between Binary Features

```python
import numpy as np

F1 = np.array([1, 1, 0, 1, 0, 0, 1, 0])
F2 = np.array([1, 0, 0, 1, 1, 0, 1, 0])

n11 = np.sum((F1 == 1) & (F2 == 1))
n00 = np.sum((F1 == 0) & (F2 == 0))
n10 = np.sum((F1 == 1) & (F2 == 0))
n01 = np.sum((F1 == 0) & (F2 == 1))

hamming = np.sum(F1 != F2)
jaccard = n11 / (n11 + n10 + n01)
jaccard_distance = 1 - jaccard
smc = (n11 + n00) / len(F1)

print("Hamming Distance:", hamming)
print("n11:", n11)
print("n00:", n00)
print("n10:", n10)
print("n01:", n01)
print("Jaccard Similarity:", jaccard)
print("Jaccard Distance:", jaccard_distance)
print("SMC:", smc)
```

### Question 5: Comparing Binary Similarity Measures

```python
import numpy as np

F1 = np.array([1, 0, 1, 1, 0, 1, 0, 0])
F2 = np.array([1, 0, 0, 1, 1, 1, 0, 1])

n11 = np.sum((F1 == 1) & (F2 == 1))
n00 = np.sum((F1 == 0) & (F2 == 0))
n10 = np.sum((F1 == 1) & (F2 == 0))
n01 = np.sum((F1 == 0) & (F2 == 1))

hamming = np.sum(F1 != F2)
jaccard_distance = 1 - n11 / (n11 + n10 + n01)
smc = (n11 + n00) / len(F1)

print("Hamming Distance:", hamming)
print("Jaccard Distance:", jaccard_distance)
print("SMC:", smc)
```

Observation: Hamming measures differences, Jaccard ignores common zeros, and SMC considers both common ones and zeros.

### Question 6: Identifying Redundant Features

```python
import pandas as pd

data = {
    'Study_Hours': [2, 4, 6, 8, 10, 12],
    'Practice_Hours': [3, 6, 9, 12, 15, 18],
    'Sleep_Hours': [8, 7, 8, 6, 7, 8],
    'Screen_Time': [2, 3, 4, 5, 6, 7]
}

df = pd.DataFrame(data)

corr = df.corr()
print(corr)

pairs = []

for i in range(len(corr.columns)):
    for j in range(i + 1, len(corr.columns)):
        if corr.iloc[i, j] > 0.95:
            pairs.append((corr.columns[i], corr.columns[j]))

print("Redundant Pairs:", pairs)

df_new = df.drop(columns=['Practice_Hours', 'Screen_Time'])
print(df_new)
```

Observation: Study_Hours, Practice_Hours and Screen_Time are highly correlated. Two features are removed.

### Question 7: Correlation Analysis of Student Performance

```python
import pandas as pd

data = {
    'Assignment': [40, 50, 60, 70, 80, 90],
    'Internal': [42, 48, 62, 68, 82, 88],
    'Attendance': [60, 65, 70, 75, 80, 85],
    'Project': [30, 45, 50, 65, 75, 85]
}

df = pd.DataFrame(data)

c1 = df['Assignment'].corr(df['Internal'])
c2 = df['Attendance'].corr(df['Project'])

print("Assignment and Internal:", c1)
print("Attendance and Project:", c2)

if c1 > c2:
    print("Assignment and Internal have stronger correlation")
else:
    print("Attendance and Project have stronger correlation")
```

### Question 8: Finding the Most Similar Feature Pair

```python
import numpy as np

data = np.array([
    [2, 4, 6, 8, 10],
    [3, 6, 9, 12, 15],
    [10, 8, 6, 4, 2],
    [4, 5, 7, 8, 9]
])

similarity = np.zeros((4, 4))

for i in range(4):
    for j in range(4):
        similarity[i, j] = np.dot(data[i], data[j]) / (
            np.linalg.norm(data[i]) * np.linalg.norm(data[j])
        )

print("Cosine Similarity Matrix:")
print(similarity)

pairs = []

for i in range(4):
    for j in range(i + 1, 4):
        pairs.append((similarity[i, j], i + 1, j + 1))

print("Most Similar Pair:", max(pairs))
```

Observation: Feature 1 and Feature 2 have the highest cosine similarity.

### Question 9: Applying a Correlation Threshold

```python
import pandas as pd

corr_data = {
    'F1': [1.00, 0.92, 0.35, 0.20],
    'F2': [0.92, 1.00, 0.40, 0.25],
    'F3': [0.35, 0.40, 1.00, 0.88],
    'F4': [0.20, 0.25, 0.88, 1.00]
}

corr_df = pd.DataFrame(corr_data, index=['F1', 'F2', 'F3', 'F4'])

threshold = 0.85

for i in range(len(corr_df.columns)):
    for j in range(i + 1, len(corr_df.columns)):
        if abs(corr_df.iloc[i, j]) > threshold:
            print(corr_df.columns[i], corr_df.columns[j])
```

### Question 12: Jaccard Similarity for Binary Features

```python
import numpy as np

Product_A = np.array([1, 1, 0, 1, 0, 1, 0, 0, 1])
Product_B = np.array([1, 0, 0, 1, 1, 1, 0, 1, 0])

n11 = np.sum((Product_A == 1) & (Product_B == 1))
n10 = np.sum((Product_A == 1) & (Product_B == 0))
n01 = np.sum((Product_A == 0) & (Product_B == 1))

jaccard = n11 / (n11 + n10 + n01)
distance = 1 - jaccard

print("Common Purchases:", n11)
print("Jaccard Similarity:", jaccard)
print("Jaccard Distance:", distance)
```

Observation: Both products have moderate similarity.

### Question 14: Calculating Minkowski Distance

```python
import numpy as np

F1 = np.array([2, 4, 6, 8])
F2 = np.array([5, 6, 8, 10])

for r in [1, 2, 3]:
    distance = np.sum(np.abs(F1 - F2) ** r) ** (1 / r)
    print("Minkowski Distance for r =", r, ":", distance)
```

Observation: r = 1 gives Manhattan distance and r = 2 gives Euclidean distance.

### Question 16: Identifying Redundancy in a Dataset

```python
import pandas as pd

data = {
    'Temperature': [20, 25, 30, 35, 40, 45],
    'Temp_Celsius': [20, 25, 30, 35, 40, 45],
    'Humidity': [60, 65, 55, 70, 50, 75],
    'Pressure': [1000, 1005, 1010, 1015, 1020, 1025]
}

df = pd.DataFrame(data)

corr = df.corr()
print("Correlation Matrix:")
print(corr)

for i in range(len(corr.columns)):
    for j in range(i + 1, len(corr.columns)):
        if abs(corr.iloc[i, j]) == 1:
            print("Perfectly Correlated:", corr.columns[i], corr.columns[j])

df_new = df.drop(columns=['Temp_Celsius'])
print("New DataFrame:")
print(df_new)
```

Observation: Temperature and Temp_Celsius are perfectly correlated. One feature is removed to avoid redundancy.
