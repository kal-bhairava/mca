## 2. **Dictionary**

Q16. Create dictionary of 5 students and print keys and values

```python
d = {
    "Aman": 78,
    "Riya": 85,
    "Kabir": 69,
    "Neha": 90,
    "Arjun": 74
}

print("Keys:", list(d.keys()))
print("Values:", list(d.values()))
```

Q17. Check whether a given key exists

```python
d = {
    "Aman": 78,
    "Riya": 85,
    "Kabir": 69,
    "Neha": 90
}

name = input("Enter name: ")

if name in d:
    print("Key found. Value =", d[name])
else:
    print("Key not found")
```

Q19. Update price of a fruit

```python
d = {
    "Apple": 100,
    "Banana": 40,
    "Mango": 60
}

fruit = input("Enter fruit: ")
price = int(input("Enter new price: "))

d[fruit] = price

print("Updated Dictionary:", d)
```

Q23. Find subject with highest marks

```python
d = {
    "Maths": 95,
    "Physics": 85,
    "English": 90
}

subject = max(d, key=d.get)

print("Highest Marks Subject:", subject, "=", d[subject])
```

Q33. Create dictionary from user input (3 pairs)

```python
d = {}

for i in range(3):
    key = input("Enter key: ")
    value = int(input("Enter value: "))
    d[key] = value

print(d)
```

Q34. Print all capitals from country-capital dictionary

Q35. Count values greater than 50

```python
d = {
    "A": 60,
    "B": 40,
    "C": 80,
    "D": 30
}

count = 0

for i in d.values():
    if i > 50:
        count += 1

print("Count =", count)
```

## 3. **Sets**

Q36. Create a set using set() from user-entered numbers

```python
L = list(map(int, input("Enter numbers: ").split()))

s = set(L)

print(s)
```

Q37. Create a set with number, decimal and word

```python
s = {10, 3.5, "Python"}

print(s)
print(len(s))
```

Q38. Convert list of cities into set

```python
cities = ["Delhi", "Mumbai", "Delhi", "Chandigarh", "Mumbai"]

s = set(cities)

print(s)
```

Q39. Create an empty set and add three elements

```python
s = set()

for i in range(3):
    n = input("Enter element: ")
    s.add(n)

print(s)
```

Q40. Add a new number to set

```python
s = {2, 4, 6}

n = int(input("Enter number: "))

s.add(n)

print(s)
```

Q41. Update a set using values from a tuple

```python
s = {1, 2, 3}
t = (4, 5, 6)

s.update(t)

print(s)
```

Q42. Update a set using characters from a string

```python
s = set()

str1 = input("Enter string: ")

s.update(str1)

print(s)
```

Q43. Merge two sets using update()

```python
s1 = {1, 2, 3}
s2 = {4, 5, 6}

s1.update(s2)

print(s1)
```

Q44. Delete an element using discard()

```python
s = {10, 20, 30, 40}

n = int(input("Enter element to delete: "))

s.discard(n)

print(s)
```

Q45. Remove all elements smaller than 10

```python
s = {2, 5, 10, 15, 20, 7}

for i in s.copy():
    if i < 10:
        s.remove(i)

print(s)
```

Q46. Check whether a word exists in programming languages set

```python
s = {"Python", "Java", "C", "C++", "PHP"}

word = input("Enter language: ")

print(word in s)
```

Q47. Remove all values divisible by 3

```python
s = {3, 6, 8, 9, 12, 15, 20}

for i in s.copy():
    if i % 3 == 0:
        s.remove(i)

print(s)
```

Q48. Display union, intersection, difference and symmetric difference

```python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

print("Union:", A | B)
print("Intersection:", A & B)
print("Difference:", A - B)
print("Symmetric Difference:", A ^ B)
```

Q51. Check whether two sets have no common elements

```python
A = {1, 2, 3}
B = {4, 5, 6}

print(A.isdisjoint(B))
```

Q52. Check subset and superset

```python
A = {1, 2}
B = {1, 2, 3, 4}

print("Subset:", A.issubset(B))
print("Superset:", A.issuperset(B))
```
