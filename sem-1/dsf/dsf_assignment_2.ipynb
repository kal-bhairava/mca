8. Display Weekday Using match-case

```python
day = int(input("Enter day number (1-7): "))

match day:
    case 1:
        print("Monday")
    case 2:
        print("Tuesday")
    case 3:
        print("Wednesday")
    case 4:
        print("Thursday")
    case 5:
        print("Friday")
    case 6:
        print("Saturday")
    case 7:
        print("Sunday")
    case _:
        print("Invalid day number")
```

11. Factorial Using while Loop

```python
n = int(input("Enter a non-negative number: "))

if n < 0:
    print("Invalid input")
else:
    fact = 1
    i = 1

    while i <= n:
        fact = fact * i
        i += 1

    print("Factorial =", fact)
```

12. Count the Digits of a Number

```python
n = int(input("Enter a number: "))

n = abs(n)
count = 0

if n == 0:
    count = 1
else:
    while n > 0:
        n = n // 10
        count += 1

print("Number of digits =", count)
```

13. Reverse a Number

```python
n = int(input("Enter a number: "))

temp = abs(n)
reverse = 0

while temp > 0:
    digit = temp % 10
    reverse = reverse * 10 + digit
    temp = temp // 10

if n < 0:
    reverse = -reverse

print("Reversed number =", reverse)
```

14. Prime Number Check

```python
n = int(input("Enter a number: "))

if n <= 1:
    print("Not Prime")
else:
    for i in range(2, n):
        if n % i == 0:
            print("Not Prime")
            break
    else:
        print("Prime")
```

17. Point Location Using match-case

```python
x = int(input("Enter x coordinate: "))
y = int(input("Enter y coordinate: "))

match (x, y):
    case (0, 0):
        print("Origin")
    case (x, 0) if x > 0:
        print("Positive X-axis")
    case (x, 0) if x < 0:
        print("Negative X-axis")
    case (0, y) if y > 0:
        print("Positive Y-axis")
    case (0, y) if y < 0:
        print("Negative Y-axis")
    case (x, y) if x > 0 and y > 0:
        print("First Quadrant")
    case (x, y) if x < 0 and y > 0:
        print("Second Quadrant")
    case (x, y) if x < 0 and y < 0:
        print("Third Quadrant")
    case (x, y) if x > 0 and y < 0:
        print("Fourth Quadrant")
```

20. Number Guessing Game Using while-else

```python
secret = 25
attempts = 0

while attempts < 5:
    guess = int(input("Guess the number: "))

    if guess == secret:
        print("Correct!")
        break
    elif guess > secret:
        print("Too High")
    else:
        print("Too Low")

    attempts += 1
else:
    print("All attempts exhausted. Game Over!")
```
